````python
# Funcione de Validación
# -----------------------------------------

# Validación de Fecha
def validar_fecha(valor):
    if pd.isna(valor):
        return False
    try:
        pd.to_datetime(valor,format="%Y-%m-%d", errors="raise")
        return True
    except:
        return False

# Validación de Hora
def validar_hora(valor):
    if pd.isna(valor):
        return False
    formatos = [
        "%H:%M",       # 13:02, 17:18, 5:50
        "%H:%M:%S",    # 13:02:00
        "%I:%M %p"     # 10:55 PM, 1:06 AM
    ]
    for formato in formatos:
        try:
            pd.to_datetime(valor, format=formato)
            return True
        except:
            pass
    return False

# Validación valores numéricos
def validar_numerico(valor):
    if pd.isna(valor):
        return False
    try:
        float(valor)
        return True
    except:
        return False

# Validación de valores texto
def validar_texto(valor):
    return not pd.isna(valor)


# -----------------------------------------
# TIPO ESPERADO DE CADA COLUMNA
# -----------------------------------------

tipos_columnas = {
    "movimiento_id": "texto",
    "buque_id": "texto",
    "matricula": "texto",
    "radar_id": "texto",
    "muelle": "texto",
    "tipo_carga": "texto",
    "origen": "texto",

    "fecha_ingreso": "fecha",
    "hora_ingreso": "hora",

    "fecha_egreso": "fecha",
    "hora_egreso": "hora",

    "tonelaje_declarado": "numerico",
    "velocidad_ingreso": "numerico",
    "velocidad_maxima_muelle": "numerico",

    "estado_despacho": "texto"
}


# -----------------------------------------
# CALCULAR COMPLETITUD
# -----------------------------------------

print("Completitud del dataset:")

for columna, tipo in tipos_columnas.items():

    if tipo == "fecha":
        correctos = df[columna].apply(validar_fecha)

    elif tipo == "hora":
        correctos = df[columna].apply(validar_hora)

    elif tipo == "numerico":
        correctos = df[columna].apply(validar_numerico)

    elif tipo == "texto":
        correctos = df[columna].apply(validar_texto)

    porcentaje = correctos.mean() * 100

    print(f" - {columna}: {porcentaje:.2f}% completa")
````

- Normalizar las horas con el formato de 24hs. Las horas inválidas deben completarse con `00:00`.

  Mostrar los datos de las 10 primeras horas inválidas.


````python
# Función para normalizar las horas
def normalizar_hora(valor):
    if pd.isna(valor):
        return "00:00"

    formatos = [
        "%H:%M",       # Ejemplo: 17:18
        "%H:%M:%S",    # Ejemplo: 17:18:30
        "%I:%M %p"     # Ejemplo: 10:55 PM
    ]

    for formato in formatos:
        try:
            hora = pd.to_datetime(valor, format=formato)
            return hora.strftime("%H:%M")
        except:
            pass

    # Si ningún formato es válido
    return "00:00"


# Guardamos las horas originales para identificar cuáles eran inválidas
horas_ingreso_original = df["hora_ingreso"].copy()
horas_egreso_original = df["hora_egreso"].copy()


# Normalizamos las dos columnas
df["hora_ingreso"] = df["hora_ingreso"].apply(normalizar_hora)
df["hora_egreso"] = df["hora_egreso"].apply(normalizar_hora)


# Identificamos las horas que eran inválidas
horas_invalidas_ingreso = horas_ingreso_original[
    df["hora_ingreso"] == "00:00"
]

horas_invalidas_egreso = horas_egreso_original[
    df["hora_egreso"] == "00:00"
]


print("Primeras 10 horas de ingreso que eran inválidas:")
print(horas_invalidas_ingreso.head(10))

print("\nPrimeras 10 horas de egreso que eran inválidas:")
print(horas_invalidas_egreso.head(10))
````


- Calcular y agregar la columna `duracion_horas`: tiempo transcurrido entre `fecha_ingreso`+`hora_ingreso` y `fecha_egreso`+`hora_egreso`. Cuando el cálculo no sea posible (fechas inválidas en alguno de los extremos) asignar `pd.NA`.

  Mostrar los datos de las 10 primeras duraciones.


````python
# Convertimos fecha y hora de ingreso en una sola fecha-hora
inicio = pd.to_datetime(
    df["fecha_ingreso"] + " " + df["hora_ingreso"],
    errors="coerce"
)

# Convertimos fecha y hora de egreso en una sola fecha-hora
fin = pd.to_datetime(
    df["fecha_egreso"] + " " + df["hora_egreso"],
    errors="coerce"
)

# Identificamos las fechas que fueron reemplazadas por 1900-01-01
fecha_ingreso_invalida = df["fecha_ingreso"] == "1900-01-01"
fecha_egreso_invalida = df["fecha_egreso"] == "1900-01-01"

# Calculamos la duración en horas
df["duracion_horas"] = (fin - inicio).dt.total_seconds() / 3600

# Cuando alguno de los extremos tenía una fecha inválida,
# asignamos pd.NA
df.loc[
    fecha_ingreso_invalida | fecha_egreso_invalida,
    "duracion_horas"
] = pd.NA

# Mostrar las 10 primeras duraciones
print("Primeras 10 duraciones:")
print(df[["fecha_ingreso", "hora_ingreso",
          "fecha_egreso", "hora_egreso",
          "duracion_horas"]].head(10))
````


- Normalizar matrículas: quitar caracteres especiales y pasar a mayúsculas. Cuando la matrícula no sea válida completar con `pd.NA`.

  Mostrar los datos de las 10 últimas.


````python
def normalizar_matricula(valor):
    # Si el valor es nulo
    if pd.isna(valor):
        return pd.NA

    # Convertir a texto y eliminar espacios al principio y al final
    valor = str(valor).strip()

    # Eliminar caracteres especiales
    # Conservamos solamente letras y números
    valor = re.sub(r'[^A-Za-z0-9]', '', valor)

    # Convertir a mayúsculas
    valor = valor.upper()

    # Si después de limpiar no quedó ningún carácter válido
    if valor == "":
        return pd.NA

    return valor


# Aplicar la normalización
df["matricula"] = df["matricula"].apply(normalizar_matricula)

# Mostrar las 10 últimas matrículas
print("Últimas 10 matrículas:")
print(df["matricula"].tail(10))
````


- Normalizar muelles: quitar caracteres especiales y pasar a mayúsculas.

````python
def normalizar_muelle(valor):
    # Si el valor es nulo, lo dejamos como pd.NA
    if pd.isna(valor):
        return pd.NA

    # Convertir a texto y quitar espacios al principio y al final
    valor = str(valor).strip()

    # Eliminar caracteres especiales
    valor = re.sub(r'[^A-Za-z0-9]', '', valor)

    # Convertir a mayúsculas
    valor = valor.upper()

    return valor


# Aplicar la normalización
df["muelle"] = df["muelle"].apply(normalizar_muelle)

# Mostrar los 10 primeros muelles
print("Primeros 10 muelles:")
print(df["muelle"].head(10))
````


- Detectar y eliminar outliers en `tonelaje_declarado` y `velocidad_ingreso`. Para esto utilizaremos los métodos [IQR](https://www.geeksforgeeks.org/machine-learning/interquartile-range-to-detect-outliers-in-data/) y [Z-score](https://www.geeksforgeeks.org/data-science/z-score-in-statistics/).

  Comparar y justificar el método elegido en un comentario dentro de la celda.

  Mostrar las columnas críticas y la cantidad de filas eliminadas para esa columna.
```
Las columnas críticas son:
 - nombre_columna: XXX
 - ...
```

````python
import numpy as np

# ============================================================
# DETECCIÓN DE OUTLIERS: IQR vs Z-SCORE
# ============================================================

columnas_outliers = [
    "tonelaje_declarado",
    "velocidad_ingreso"
]

# ------------------------------------------------------------
# MÉTODO IQR
# ------------------------------------------------------------

outliers_iqr = {}

for columna in columnas_outliers:

    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    outliers_iqr[columna] = (
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    )


# ------------------------------------------------------------
# MÉTODO Z-SCORE
# ------------------------------------------------------------

outliers_zscore = {}

for columna in columnas_outliers:

    media = df[columna].mean()
    desvio = df[columna].std()

    z_score = (df[columna] - media) / desvio

    outliers_zscore[columna] = abs(z_score) > 3


# ------------------------------------------------------------
# COMPARACIÓN DE MÉTODOS
# ------------------------------------------------------------

print("COMPARACIÓN DE MÉTODOS")
print("-" * 50)

for columna in columnas_outliers:

    cantidad_iqr = outliers_iqr[columna].sum()
    cantidad_zscore = outliers_zscore[columna].sum()

    print(f"{columna}:")
    print(f"  IQR:     {cantidad_iqr} outliers")
    print(f"  Z-score: {cantidad_zscore} outliers")
    print()


# ============================================================
# MÉTODO ELEGIDO: IQR
# ============================================================

# Se elige IQR porque es más robusto frente a valores extremos
# y no requiere asumir que los datos tienen una distribución
# normal. Por este motivo resulta más apropiado para este
# dataset, especialmente para variables como el tonelaje,
# donde pueden existir valores muy alejados del resto.
#
# El método Z-score resulta útil cuando los datos presentan
# una distribución aproximadamente normal, pero puede verse
# afectado por los propios valores extremos al calcular la
# media y el desvío estándar.

filas_outliers = (
    outliers_iqr["tonelaje_declarado"] |
    outliers_iqr["velocidad_ingreso"]
)

# Cantidad de filas que se eliminarán
total_eliminadas = filas_outliers.sum()

# Cantidad de filas eliminadas atribuida a cada columna
eliminadas_tonelaje = outliers_iqr["tonelaje_declarado"].sum()
eliminadas_velocidad = outliers_iqr["velocidad_ingreso"].sum()

print("RESULTADO DEL MÉTODO IQR")
print("-" * 50)

print(f" - tonelaje_declarado: {eliminadas_tonelaje} filas con outliers")
print(f" - velocidad_ingreso: {eliminadas_velocidad} filas con outliers")
print(f"\nTotal de filas eliminadas: {total_eliminadas}")


# ------------------------------------------------------------
# ELIMINAR OUTLIERS
# ------------------------------------------------------------

df = df.loc[~filas_outliers].copy()

print(f"Filas restantes en el dataset: {len(df)}")
````
